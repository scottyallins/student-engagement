```python
CATALOG = "crm_ingestion"

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.config
""")

print("✅ Config schema created")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS crm_ingestion.config.bronze_config (
    source_table STRING,
    target_table STRING,
    incremental_column STRING,
    load_type STRING,
    active BOOLEAN
)
USING DELTA
""")

print("✅ Bronze config table created")
```

```python
spark.sql("""
CREATE OR REPLACE TABLE crm_ingestion.config.bronze_config
AS

SELECT * FROM VALUES

('raw.all_payments',
 'crm_ingestion.bronze.all_payments',
 'insert_date',
 'incremental',
 true),

('raw.calendly_scheduled_events',
 'crm_ingestion.bronze.calendly_scheduled_events',
 'insert_date',
 'incremental',
 true),

('raw.close_crm_users_raw',
 'crm_ingestion.bronze.close_crm_users_raw',
 'insert_date',
 'incremental',
 true),

('raw.custom_activites_raw',
 'crm_ingestion.bronze.custom_activites_raw',
 'insert_date',
 'incremental',
 true),

('raw.lead_activites_raw',
 'crm_ingestion.bronze.lead_activites_raw',
 'insert_date',
 'incremental',
 true),

('raw.lead_merges',
 'crm_ingestion.bronze.lead_merges',
 'insert_date',
 'incremental',
 true),

('raw.leads_raw',
 'crm_ingestion.bronze.leads_raw',
 'insert_date',
 'incremental',
 true),

('raw.mdl_users_raw',
 'crm_ingestion.bronze.mdl_users_raw',
 'insert_date',
 'incremental',
 true),

('raw.student_sentiment',
 'crm_ingestion.bronze.student_sentiment',
 'insert_date',
 'incremental',
 true)

AS t (
 source_table,
 target_table,
 incremental_column,
 load_type,
 active
)
""")
```

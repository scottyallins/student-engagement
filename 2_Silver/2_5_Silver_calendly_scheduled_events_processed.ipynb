```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

df = spark.table('crm_ingestion.silver.calendly_scheduled_events_processed')

window = Window.partitionBy("EVENT_TYPE_NAME").orderBy(df.EVENT_START_TIME.desc())

display(
    df.withColumn("rn", row_number().over(window))
    .filter("rn = 1")
    .drop("rn")
    .sort('EVENT_START_TIME', ascending=False)
    .limit(10)
)
```
